```python
import pandas as pd
import numpy as np

df = pd.read_parquet('../Data/traffic_feature_engineered.parquet')

print(df.shape)
df.head()
```

```python
df = df.sort_values('date_time').reset_index(drop=True)
```

```python
features = [
    'temp',
    'rain_1h',
    'snow_1h',
    'clouds_all',
    'fog_mm',
    'wind_speed_ms',
    'flood_mm',
    'hour',
    'day_of_week',
    'month',
    'year',
    'is_weekend',
    'is_holiday',
    'holiday',
    'weather_description',
    'time_period'
]
```

```python
target = 'traffic_volume'
```

```python
X = df[features]
y = df[target]
```

```python
train_mask = df['year'] < 2018
test_mask = df['year'] == 2018
```

```python
X_train = X[train_mask]
X_test = X[test_mask]

y_train = y[train_mask]
y_test = y[test_mask]
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
numeric_features = [
    'temp',
    'rain_1h',
    'snow_1h',
    'clouds_all',
    'fog_mm',
    'wind_speed_ms',
    'flood_mm',
    'hour',
    'day_of_week',
    'month',
    'year',
    'is_weekend',
    'is_holiday'
]

categorical_features = [
    'holiday',
    'weather_description',
    'time_period'
]
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)
```

```python
import joblib

joblib.dump(
    preprocessor,
    '../Models/regression_preprocessor.pkl'
)
```

```python
joblib.dump(X_train_processed, '../Data/X_train_reg.pkl')
joblib.dump(X_test_processed, '../Data/X_test_reg.pkl')

joblib.dump(y_train, '../Data/y_train_reg.pkl')
joblib.dump(y_test, '../Data/y_test_reg.pkl')
```

```python
print("Regression Data")
print("X_train:", X_train_processed.shape)
print("X_test:", X_test_processed.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```
